## Array orient Programming

#### Graph Plotting

Using NumPy arrays enables you to express many kinds of data processing tasks as
concise array expressions that might otherwise require writing loops.

This practice
of replacing explicit loops with array expressions is referred to by some people
as vectorization.

```python
# plotting a 2D plot
# x = y
import matplotlib.pyplot as plt
import numpy as np 

x = np.linspace(-10,10,100)
y = x

plt.plot(x,y)
```

```python
# y = x^2
x = np.linspace(-10,10,100)
y = x**2

plt.plot(x,y)
```

```python
# y = sin(x)
x = np.linspace(-10,10,100)
y = np.sin(x)

plt.plot(x,y)
```

```python
# y = xlog(x)
x = np.linspace(-10,10,100)
y = x * np.log(x)

plt.plot(x,y)
```

```python
# sigmoid
x = np.linspace(-10,10,100)
y = 1/(1+np.exp(-x))

plt.plot(x,y)
```

## Expressing Conditional Logic as Array Operations

np.where(condition, value_if_true, value_if_false)

- Iska matlab:

- condition: kya check karna hai?

- value_if_true: agar condition True ho to kya value leni hai?

- value_if_false: agar condition False ho to kya value leni hai?

```python
import numpy as np

xarr = np.array([1.1, 1.2, 1.3, 1.4, 1.5])
yarr = np.array([2.1, 2.2, 2.3, 2.4, 2.5])

cond = np.array([True, False, True, True, False])
```

| Index | `cond` | `xarr` | `yarr` | Result |
|---:|:---:|---:|---:|---:|
| 0 | True  | 1.1 | 2.1 | 1.1 |
| 1 | False | 1.2 | 2.2 | 2.2 |
| 2 | True  | 1.3 | 2.3 | 1.3 |
| 3 | True  | 1.4 | 2.4 | 1.4 |
| 4 | False | 1.5 | 2.5 | 2.5 |

```python
result = np.where(cond, xarr, yarr)

print(result)
```

```python
import numpy as np

rng = np.random.default_rng(42)

arr = np.random.standard_normal((4, 4))

print(arr)
```

```python
print(arr > 0)
```

```python
result = np.where(arr > 0, 2, -2)

print(result)
```

```python
result = np.where(arr > 0, 2, arr)
result
```

```python
arr = np.array([5, -3, 8, -1, 0])

result = np.where(arr > 0, 2, arr)

print(result)
```

| Original Value | Result |
|---:|---:|
| 5  | 2  |
| -3 | -3 |
| 8  | 2  |
| -1 | -1 |
| 0  | 0  |

```python
arr = np.array([10, -5, 20, -8])

result = np.where(arr > 0, 100, arr)

print(result)
```

| Original Value | `arr > 0` | Kya hoga? | Result |
|---:|:---:|---|---:|
| 10  | True  | 100 se replace | 100 |
| -5  | False | Original value rakho | -5 |
| 20  | True  | 100 se replace | 100 |
| -8  | False | Original value rakho | -8 |

```python
arr = np.array([12, -4, 0, 7, -9, 3])

result = np.where(arr >= 0, arr + 2, -1)

print(result)
```

## Mathematical and Statistical Methods

### Basic Array Statistical Methods

| Method | Description |
|---|---|
| `sum` | Sum of all the elements in the array or along an axis; zero-length arrays have sum `0` |
| `mean` | Arithmetic mean; invalid (returns `NaN`) on zero-length arrays |
| `std`, `var` | Standard deviation and variance, respectively |
| `min`, `max` | Minimum and maximum values |
| `argmin`, `argmax` | Indices of minimum and maximum elements, respectively |
| `cumsum` | Cumulative sum of elements starting from `0` |
| `cumprod` | Cumulative product of elements starting from `1` |

### Table Array Concatenation Functions

| Function | Description |
|---|---|
| `concatenate` | Most general function; concatenate a collection of arrays along one axis |
| `vstack`, `row_stack` | Stack arrays by rows (along axis 0) |
| `hstack` | Stack arrays by columns (along axis 1) |
| `column_stack` | Like `hstack`, but convert 1D arrays to 2D column vectors first |
| `dstack` | Stack arrays by depth (along axis 2) |
| `split` | Split an array at passed locations along a particular axis |
| `hsplit`, `vsplit` | Convenience functions for splitting along axis 1 and axis 0, respectively |
