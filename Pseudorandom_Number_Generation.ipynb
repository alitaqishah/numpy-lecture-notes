## Pseudorandom Number Generation

The numpy.random module supplements the built-in Python random module with
functions for efficiently generating whole arrays of sample values from many kinds of
probability distributions.
- Pseudo-random = random jaisa lagne wala number, jo actually ek algorithm se generate hua hai.

### Table. NumPy Random Number Generator Methods

| Method | Description |
|---|---|
| `permutation` | Return a random permutation of a sequence, or return a permuted range |
| `shuffle` | Randomly permute a sequence in place |
| `uniform` | Draw samples from a uniform distribution |
| `integers` | Draw random integers from a given low-to-high range |
| `standard_normal` | Draw samples from a normal distribution with mean 0 and standard deviation 1 |
| `binomial` | Draw samples from a binomial distribution |
| `normal` | Draw samples from a normal (Gaussian) distribution |
| `beta` | Draw samples from a beta distribution |
| `chisquare` | Draw samples from a chi-square distribution |
| `gamma` | Draw samples from a gamma distribution |
| `uniform` | Draw samples from a uniform `[0, 1)` distribution |

```python
arr = np.arange(1, 6)

result = np.random.permutation(arr)

print(result)
```

```python
arr = np.arange(1, 6)

result = rng.permutation(arr)

print(result)
```

```python
arr = np.arange(1, 6)

rng.shuffle(arr)

print(arr)
```

```python
arr = np.arange(1, 6)

np.random.shuffle(arr)

print(arr)
```

```python
numbers = rng.integers(1, 10, size=5)

print(numbers)
```

| Function            | Kya generate karta hai?                 | Example               |
|---------------------|-----------------------------------------|-----------------------|
| `permutation()`     | Randomly rearranged sequence            | `[3,1,5,2,4]`         |
| `shuffle()`         | Sequence ko in-place rearrange          | original array changes |
| `integers()`        | Random integers                         | `1, 7, 3, 9`          |
| `standard_normal()` | Normal distribution, mean=0, SD=1       | `-0.4, 1.2`            |
| `normal()`          | Normal distribution with chosen mean/SD | mean=50, SD=10        |
| `binomial()`        | Number of successes                     | `6 heads out of 10`   |
| `beta()`            | Beta-distributed values                 | `0.18, 0.42`          |
| `chisquare()`       | Chi-square distributed values           | `3.2, 6.5`            |
| `gamma()`           | Gamma-distributed values                | `2.3, 5.7`            |
| `uniform()`         | Uniform random values                   | `0.13, 0.72`          |
