## Broadcasting

- The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations.

- The smaller array is “broadcast” across the larger array so that they have compatible shapes.

```python
import numpy as np

marks = np.array([60, 70, 80, 90, 100])
```

```python
marks + 5
```

```python
# same shape
a = np.arange(6).reshape(2,3)
b = np.arange(6,12).reshape(2,3)

print(a)


```

```python
print(b)
```

```python
print(a+b)
```

```python
# diff shape
a = np.arange(6).reshape(2,3)
b = np.arange(3).reshape(1,3)
```

```python
print(a)
```

```python
print(b)
```

```python
print(a+b)
```

## Broadcasting ke basic examples
### Example A: Scalar + 1D array

```python
a = np.array([10, 20, 30])

print(a + 2)
print(a * 3)
```

### Example B: Same-size arrays

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print(a + b)
#Yahan broadcasting ki zaroorat bhi nahi, kyunke dono arrays ki shape same hai: (3,).
```

### Example C: Different-size arrays

```python
a = np.array([10, 20, 30])
b = np.array([5])

print(a + b)

#b ki shape (1,) hai aur a ki (3,). NumPy single value ko compatible samajh kar har element ke saath use karta hai.
```

Broadcasting Rules
1. Make the two arrays have the same number of dimensions.

- If the numbers of dimensions of the two arrays are different, add new dimensions with size 1 to the head of the array with the smaller dimension.
2. Make each dimension of the two arrays the same size.

- If the sizes of each dimension of the two arrays do not match, dimensions with size 1 are stretched to the size of the other array.
If there is a dimension whose size is not 1 in either of the two arrays, it cannot be broadcasted, and an error is raised.

NumPy shapes ko kaise compare karta hai?

- Broadcasting ke liye NumPy shapes ko right side se compare karta hai.

- Do dimensions compatible hoti hain agar:

- Dono dimensions equal hon, ya

- Un mein se koi ek dimension 1 ho.

- Agar dono conditions mein se koi bhi poori na ho, broadcasting fail ho jati hai.

2D array and 1D array — row-wise broadcasting

```python
marks = np.array([
    [60, 70, 80],
    [40, 50, 60]
])

bonus = np.array([5, 10])

print(marks.shape)
print(bonus.shape)
```

```python
print(marks + bonus)
```

Column-wise broadcasting

```python
marks = np.array([
    [60, 70, 80],
    [40, 50, 60]
])

bonus = np.array([
    [5],
    [10]
])

print(bonus.shape)
print(marks + bonus)
```

## Broadcasting in Machine Learning

```python
X = np.array([
    [10, 100],
    [20, 200],
    [30, 300]
])
print(X.shape)
```

```python
mean = X.mean(axis=0)
print(mean)
```

```python
X_centered = X - mean
print(X_centered)
```

- Is process ko mean-centering kehte hain. Data preprocessing aur kai machine-learning algorithms mein yeh useful hai.
