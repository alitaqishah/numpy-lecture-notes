## Fancy Indexing

- Fancy indexing ka matlab hai ke hum index numbers ki list/array de kar apni marzi ke elements select karte hain.

```python
arr = np.arange(10,21)
arr
```

```python
#arr[start, stop, step]
arr[0:10:2]
```

```python
arr[[2, 5, 8]]
```

```python
arr[2:8] #means continuous range
arr[[2, 5, 8]] # means non-continuous range/means specific positions:
# fancy indexing can also be called as pick and choose indexing 
```

```python
arr = np.arange(12).reshape(3, 4)
arr
```

| Row \ Column |  0 |  1 |  2 |  3 |
| ------------ | -: | -: | -: | -: |
| 0            |  0 |  1 |  2 |  3 |

| 2            |  8 |  9 | 10 | 11 |


```python
arr[[0, 2]]
```

| ------------ | -: | -: | -: | -: |

```python
arr = np.arange(32).reshape((8, 4))
arr
```

```python
arr[[1, 5, 7, 2]]
```

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]]
#Here the elements (1, 0), (5, 3), (7, 1), and (2, 2) were selected
```

```python
arr[[1, 5, 7, 2]][:, [0, 3, 1, 2]]
```

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]]
```

```python
arr[[1, 5, 7, 2], [0, 3, 1, 2]] = 0
arr
```

## Boolean indexing
- Boolean indexing ka matlab hai True / False ki condition ke basis par array se values select karna.

```python
import numpy as np

arr = np.arange(10, 70, 10)

print(arr)
```

```python
# pehli condition
arr > 30
```

| Value | Condition | Result |
|---:|:---:|:---:|
| 10 | 10 > 30 | False |
| 20 | 20 > 30 | False |
| 30 | 30 > 30 | False |
| 40 | 40 > 30 | True |
| 50 | 50 > 30 | True |
| 60 | 60 > 30 | True |

```python
#Is Boolean array ko original array mein lagao
arr[arr > 30]
```

| Index | 0  | 1  | 2  | 3  | 4  | 5  |
|---|---:|---:|---:|---:|---:|---:|
| **Original** | 10 | 20 | 30 | 40 | 50 | 60 |
| **Condition** | F | F | F | T | T | T |
| **Keep?** | NO | NO | NO | YES | YES | YES |

**Result:** `[40, 50, 60]`

```python
arr < 40
```

```python
arr[arr < 40]
```

```python
arr[arr == 30]
```

```python
arr != 30
```

```python
arr[arr != 30]
```

```python
arr = np.arange(1, 13).reshape(3, 4)

print(arr)
```

```python
arr > 6
```

| Value | Condition | Result |
|---:|:---:|:---:|
| 1  | 1 > 6  | False |
| 2  | 2 > 6  | False |
| 3  | 3 > 6  | False |
| 4  | 4 > 6  | False |
| 5  | 5 > 6  | False |
| 6  | 6 > 6  | False |
| 7  | 7 > 6  | True |
| 8  | 8 > 6  | True |
| 9  | 9 > 6  | True |
| 10 | 10 > 6 | True |
| 11 | 11 > 6 | True |
| 12 | 12 > 6 | True |

```python
arr[arr > 6]
```

```python
arr = np.arange(10, 70, 10)
```

```python
#Values jo 20 se greater AND 50 se less
arr[(arr > 20) & (arr < 50)]
```

```python
#Values jo 20 se less OR 50 se greater hain.
arr[(arr < 20) | (arr > 50)]
```

```python
arr > 30
```

```python
~ (arr > 30)
```

```python
#not
arr[~(arr > 30)]
```

```python
a = np.random.randint(1,100,24).reshape(6,4)
a
```

```python
a > 50
```

```python
# find all numbers greater than 50
a[a > 50]
```

```python
# find out even numbers
a[a % 2 == 0]
```

```python
# find all numbers greater than 50 and are even

a[(a > 50) & (a % 2 == 0)]
```

```python
# find all numbers not divisible by 7
a[~(a % 7 == 0)]
```
