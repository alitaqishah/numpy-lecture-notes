```python
import numpy as np
x = np.array([[1., 2., 3.], [4., 5., 6.]])
y = np.array([[6., 23.], [-1, 7], [8, 9]])
```

```python
x
```

```python
y
```

```python
x.dot(y)
```

```python
np.dot(x, y)
```

```python
x @ np.ones(3)
```

```python
from numpy.linalg import inv, qr
X = np.random.standard_normal((5, 5))
mat = X.T @ X
inv(mat)
mat @ inv(mat)
```

### Table 4-8. Commonly Used `numpy.linalg` Functions

| Function | Description |
|---|---|
| `diag` | Return the diagonal (or off-diagonal) elements of a square matrix as a 1D array, or convert a 1D array into a square matrix with zeros on the off-diagonal |
| `dot` | Matrix multiplication |
| `trace` | Compute the sum of the diagonal elements |
| `det` | Compute the matrix determinant |
| `eig` | Compute the eigenvalues and eigenvectors of a square matrix |
| `inv` | Compute the inverse of a square matrix |
| `pinv` | Compute the Moore-Penrose pseudoinverse of a matrix |
| `qr` | Compute the QR decomposition |
| `svd` | Compute the singular value decomposition (SVD) |
| `solve` | Solve the linear system \(Ax = b\) for \(x\), where \(A\) is a square matrix |
| `lstsq` | Compute the least-squares solution to \(Ax = b\) |

# Element-wise multiplication * vs matrix multiplication @

```python
import numpy as np

x = np.array([
    [1, 2],
    [3, 4]
])

y = np.array([
    [5, 6],
    [7, 8]
])
```

```python
print(x + y)
```

```python
print(x * y)
```

```python
print(x. dot( y))
```

```python
print(x.dot(y))
```

```python
np.dot(x, y)
```

```python
x @ y
```

# Transpose

```python
X = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(X.T)
```

```python
print(X.shape)    # (2, 3)
```

```python
print(X.T.shape)  # (3, 2)
```

```python
X.T.dot(X)
```

```python
X.T @ X
```

# Matrix inverse

```python
from numpy.linalg import inv

A = np.array([
    [2., 1.],
    [1., 2.]
])

A_inv = inv(A)

print(A_inv)
print(A @ A_inv)
```

| Function | Kaam |
|---|---|
| `np.diag(A)` | Diagonal values nikalna; 1D array se diagonal matrix banana bhi mumkin hai |
| `np.dot(A, B)` | Dot product; 2D arrays ke liye matrix multiplication |
| `np.trace(A)` | Diagonal values ka sum |
| `np.linalg.det(A)` | Determinant calculate karna |
| `np.linalg.eig(A)` | Eigenvalues aur eigenvectors nikalna |
| `np.linalg.inv(A)` | Matrix inverse nikalna |
| `np.linalg.pinv(A)` | Moore–Penrose pseudoinverse nikalna |
| `np.linalg.qr(A)` | QR decomposition karna |
| `np.linalg.svd(A)` | Singular value decomposition karna |
| `np.linalg.solve(A, b)` | Linear equations `Ax = b` solve karna |
| `np.linalg.lstsq(A, b, rcond=None)` | Least-squares solution nikalna |
