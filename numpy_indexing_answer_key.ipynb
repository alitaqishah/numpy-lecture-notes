## Part A: Predict the Output

### A1. Negative Indexing
What is the output of the following code?

```python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[-1], arr[-4])
```

**Answer:**

```
60 30
```

**Explanation:** Negative indices count from the end: -1 is the last element (60) and -4 is the fourth from the end (30).

````python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[-1], arr[-4])
````

### A2. 1D Slicing with a Step
What is the output of the following code?

```python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[1:5:2])
```

**Answer:**

```
[20 40]
```

**Explanation:** The slice starts at index 1, stops before index 5, and moves in steps of 2, selecting indices 1 and 3.

````python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[1:5:2])
````

### A3. Reverse Slicing
What is the output of the following code?

```python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[::-2])
```

**Answer:**

```
[60 40 20]
```

**Explanation:** A negative step traverses the array from the end, taking every second element: indices 5, 3 and 1.

````python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[::-2])
````

### A4. 2D Indexing
What is the output of the following code?

```python
import numpy as np
a = np.arange(1, 13).reshape(3, 4)
print(a[2, 1])
print(a[:, 2])
```

**Answer:**

```
10
[ 3  7 11]
```

**Explanation:** `a[2, 1]` selects row 2, column 1. `a[:, 2]` selects all rows of column 2.

````python
import numpy as np
a = np.arange(1, 13).reshape(3, 4)
print(a[2, 1])
print(a[:, 2])
````

### A6. 3D Indexing
What is the output of the following code?

```python

**Answer:**

```

### A12. Boolean Indexing with Two Conditions
What is the output of the following code?

```python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[(arr > 20) & (arr < 60)])
```

**Answer:**

```
[30 40 50]
```

**Explanation:** `&` combines both conditions. Only values strictly greater than 20 and strictly less than 60 are kept.

````python
import numpy as np
arr = np.arange(10, 70, 10)
print(arr[(arr > 20) & (arr < 60)])
````

### A13. Boolean Indexing on a 2D Array
What is the output of the following code?

```python
import numpy as np
a = np.arange(1, 13).reshape(3, 4)
print(a[a % 3 == 0])
print(a[a > 6].shape)
```

**Answer:**

```
[ 3  6  9 12]
(6,)
```

**Explanation:** Boolean indexing on a 2D array returns a 1D array of the selected elements, so the shape of the second result is (6,).

````python
import numpy as np
a = np.arange(1, 13).reshape(3, 4)
print(a[a % 3 == 0])
print(a[a > 6].shape)
````

### B11. Boolean Indexing in 1D
Given the marks `[34, 78, 56, 90, 45, 67, 23, 88]`, print (a) all marks above 60, and (b) all marks between 40 and 80 (inclusive).

**Expected output:**

```
[78 90 67 88]
[78 56 45 67]
```

````python
import numpy as np
marks = np.array([34, 78, 56, 90, 45, 67, 23, 88])
print(marks[marks > 60])
print(marks[(marks >= 40) & (marks <= 80)])
````

### B12. Boolean Indexing in 2D
Create a (4, 5) array with the numbers 1 to 20. Print (a) all numbers divisible by 3, (b) all even numbers greater than 10, and (c) all numbers not divisible by 5.

**Expected output:**

```
[ 3  6  9 12 15 18]
[12 14 16 18 20]
[ 1  2  3  4  6  7  8  9 11 12 13 14 16 17 18 19]
```

````python
import numpy as np
a = np.arange(1, 21).reshape(4, 5)
print(a[a % 3 == 0])
print(a[(a % 2 == 0) & (a > 10)])
print(a[~(a % 5 == 0)])
````
