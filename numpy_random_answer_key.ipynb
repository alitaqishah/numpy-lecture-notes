### A4. Successive Calls on One Generator

### A5. permutation() and the Original Array
What is the output of the following code?

```python
import numpy as np
rng = np.random.default_rng(seed=0)

```

**Answer:**

```

```

**Explanation:** `permutation()` returns a new shuffled array and leaves the original unchanged. Sorting the result recovers the values 1 to 5.

````python
import numpy as np
rng = np.random.default_rng(seed=0)
arr = np.arange(1, 6)
result = rng.permutation(arr)
print(arr)
print(np.sort(result))
````

### A14. Value Ranges of Different Distributions
What is the output of the following code?

```python

rng = np.random.default_rng(seed=21)
b = rng.beta(a=2, b=5, size=500)
```

### A15. Python list vs NumPy Array
What is the output of the following code?

```python
import numpy as np
from random import normalvariate
x = [normalvariate(0, 1) for _ in range(5)]
y = np.random.standard_normal(5)
print(type(x), type(y))
```

**Answer:**

```
<class 'list'> <class 'numpy.ndarray'>
```

**Explanation:** The list comprehension with `normalvariate` builds a Python list, while NumPy returns an `ndarray` directly. This is also why NumPy is much faster for large samples.

````python
import numpy as np
from random import normalvariate
x = [normalvariate(0, 1) for _ in range(5)]
y = np.random.standard_normal(5)
print(type(x), type(y))
````
